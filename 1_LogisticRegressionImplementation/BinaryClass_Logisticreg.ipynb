```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import warnings
warnings.filterwarnings('ignore')

%matplotlib inline
```

```python
# We are not use any external data set for this example we create our own data set
from sklearn.datasets import make_classification
```

```python
# create dataset since we are solving binary classification problem so we keep n of class is 2 
X,y = make_classification(n_samples=1000, n_features=10, n_classes=2, n_informative=2, n_redundant=0, n_clusters_per_class=1, random_state=42)
```

```python
pd.DataFrame(X)
```

```python
y
```

```python
from sklearn.model_selection import train_test_split
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.30, random_state=42)
```

```python
# model training
from sklearn.linear_model import LogisticRegression
logistic_model = LogisticRegression()
logistic_model.fit(X_train, y_train)
```

```python
# model prediction
y_pred = logistic_model.predict(X_test)
print(y_pred)
```

```python

from sklearn.metrics import accuracy_score
accuracy_score(y_test, y_pred)


```

```python
from sklearn.metrics import confusion_matrix, classification_report,roc_auc_score,roc_curve
cm = confusion_matrix(y_test, y_pred)
print(cm)

print(classification_report(y_test, y_pred))

```

```python
# to check the roc curve - how good the model is
roc_auc_score(y_test, y_pred)

fpr, tpr, thresholds = roc_curve(y_test, y_pred)
plt.plot(fpr, tpr)
plt.xlabel('False Positive Rate')
plt.ylabel('True Positive Rate')
plt.title('ROC Curve')
plt.show()
```

```python
# to check the log loss - to chck quality of the model
from sklearn.metrics import log_loss
log_loss(y_test, y_pred)


```

#### Hyper Parameter Tunning and Cross validation


****GridSearchCV performs an exhaustive search over all specified hyperparameter combinations, while RandomizedSearchCV samples a fixed number of combinations randomly. GridSearch is suitable for smaller search spaces, whereas RandomizedSearch is generally more efficient for large search spaces.****

```python
# Alwayg check out logistic regresion docs for different different parameter 
# https://scikit-learn.org/stable/modules/generated/sklearn.linear_model.LogisticRegression.html

model = LogisticRegression()
penalty = ['l1', 'l2', 'elasticnet']
C_values = [0.001, 0.01, 0.1, 1, 10, 100]
solver = ['newton-cg', 'lbfgs', 'liblinear', 'sag', 'saga']
```

```python
# make all parmse into key value pair
params = dict(penalty=penalty, C=C_values, solver=solver)


```

```python
# we decide Cross validation
from sklearn.model_selection import StratifiedKFold
cv = StratifiedKFold(n_splits=5, shuffle=True, random_state=42)
```

### Grid SearchCV

```python
# Grid Search Hyperparameter Optimization
from sklearn.model_selection import GridSearchCV

grid = GridSearchCV(estimator=model, param_grid=params, scoring='accuracy', cv=cv, n_jobs=-1)
grid.fit(X_train, y_train)



```

```python
grid.fit(X_train, y_train)
```

```python
# we can check best parameter and best score
grid.best_params_
```

```python
grid.best_score_
```

### RandomizeSearchCV

```python
from sklearn.model_selection import RandomizedSearchCV  
model = LogisticRegression()
penalty = ['l1', 'l2', 'elasticnet']
solver = ['newton-cg', 'lbfgs', 'liblinear', 'sag', 'saga']


```

```python
randomcv = RandomizedSearchCV(estimator=model, param_distributions=params, scoring='accuracy', cv=cv, n_jobs=-1)
randomcv.fit(X_train, y_train)
```

```python
randomcv.best_score_
```

```python
randomcv.best_params_
```

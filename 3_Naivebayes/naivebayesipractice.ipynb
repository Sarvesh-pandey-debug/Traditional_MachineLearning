```python
import seaborn as sns
df = sns.load_dataset('tips')
```

```python
df.info()
```

```python
# to check how many day count are there in dataset
df['day'].value_counts()
```

```python
### dependent and independent features 
X = df[['total_bill','tip', 'sex', 'smoker',  'time', 'size']]
y = df['day']
```

```python
## train test split
from sklearn.model_selection import train_test_split
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.25, random_state=0)
```

```python
X_train
```

```python
y
```

```python
## In this case we are  pickup all the features which are binary
#  we going to apply lable encoding and excluding binary apply One Hot Encoding

from sklearn.preprocessing import LabelEncoder
import warnings
warnings.filterwarnings('ignore') # to hide pandas SettingWithCopy warnings
```

```python
le1 = LabelEncoder()
le2 = LabelEncoder()
le3 = LabelEncoder()
```

```python
# 1. Transform X_train (using fit_transform)
X_train['sex'] = le1.fit_transform(X_train['sex'])
X_train['smoker'] = le2.fit_transform(X_train['smoker'])
X_train['time'] = le3.fit_transform(X_train['time'])

# 2. Transform X_test (using transform ONLY to prevent data leakage)
X_test['sex'] = le1.transform(X_test['sex'])
X_test['smoker'] = le2.transform(X_test['smoker'])
X_test['time'] = le3.transform(X_test['time'])

```

```python
X_train.head()
```

```python
X_test.head()
```

```python
y
```

```python
# For the target variable 'day' (y), we use Label Encoding
le_y = LabelEncoder()

# Use fit_transform on the training set
y_train = le_y.fit_transform(y_train)

# Use transform ONLY on the testing set
y_test = le_y.transform(y_test)
```

```python
y_test
```

```python
##1. Train the Naive Bayes Model
from sklearn.naive_bayes import GaussianNB

# Initialize the model
gnb = GaussianNB()

# Train the model on the training data
gnb.fit(X_train, y_train)
```

```python
# Predict the output on the testing data
y_pred = gnb.predict(X_test)
```

```python
y_pred
```

```python
from sklearn.metrics import accuracy_score, classification_report, confusion_matrix
```

```python
print(confusion_matrix(y_pred,y_test))
print(accuracy_score(y_pred,y_test))
print(classification_report(y_pred,y_test))
```

```python
import pandas as pd

print("--- 📝 Enter Customer Details ---")

try:
    # 1. Ask the user for inputs at runtime
    total_bill = float(input("Enter Total Bill (e.g. 25.50): "))
    tip = float(input("Enter Tip Amount (e.g. 5.00): "))
    size = int(input("Table Size (number of people, e.g. 2): "))
    
    # We use .strip().capitalize() so that if you type "male" or " male ", it fixes it to "Male"
    sex = input("Enter Sex (Male/Female): ").strip().capitalize()
    smoker = input("Smoker? (Yes/No): ").strip().capitalize()
    time = input("Time of meal (Lunch/Dinner): ").strip().capitalize()

    # 2. Package it into a DataFrame exactly how the model expects it
    user_input = pd.DataFrame({
        'total_bill': [total_bill],
        'tip': [tip],
        'sex': [sex],
        'smoker': [smoker],
        'time': [time],
        'size': [size]
    })

    # 3. Encode the text inputs into numbers using your existing LabelEncoders
    user_input['sex'] = le1.transform(user_input['sex'])
    user_input['smoker'] = le2.transform(user_input['smoker'])
    user_input['time'] = le3.transform(user_input['time'])

    # 4. Make the Prediction
    predicted_numeric_day = gnb.predict(user_input)
    
    # Convert the numeric prediction back to text ('Sun', 'Thur', etc.)
    predicted_text_day = le_y.inverse_transform(predicted_numeric_day)

    print("\n" + "="*45)
    print(f"🤖 The model predicts this meal happened on: {predicted_text_day[0]}")
    print("="*45)

except ValueError:
    print("\n❌ Error: Please make sure you enter numbers for the Bill, Tip, and Size!")
except Exception as e:
    print(f"\n❌ Error: Please check your spelling for the text inputs. Details: {e}")
```
